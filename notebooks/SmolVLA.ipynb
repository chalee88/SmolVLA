```python
import os
import sys
import json
import subprocess
from pathlib import Path
from datetime import datetime

suites = [
    "libero_spatial",
    "libero_object",
    "libero_goal",
    "libero_10",
]

episodes_per_task = 10

root_dir = (
    Path("outputs/eval")
    / datetime.now().strftime("smolvla_400_%Y%m%d_%H%M%S")
)
root_dir.mkdir(parents=True, exist_ok=True)

env = os.environ.copy()
env["MUJOCO_GL"] = "egl"
env["PYOPENGL_PLATFORM"] = "egl"
env["PYTHONUNBUFFERED"] = "1"

# Free the notebook's model copy: the evaluator loads its own.
import gc
import torch

if "policy" in globals():
    del policy
gc.collect()
torch.cuda.empty_cache()

summary = {}

for suite in suites:
    output_dir = root_dir / suite
    output_dir.mkdir(parents=True, exist_ok=True)
    log_path = output_dir / "evaluation.log"

    print(f"\nStarting {suite}: 10 tasks × 10 episodes = 100 attempts",
          flush=True)
    print(f"Log: {log_path.resolve()}", flush=True)

    command = [
        sys.executable,
        "-m", "lerobot.scripts.lerobot_eval",
        "--policy.path=HuggingFaceVLA/smolvla_libero",
        "--policy.device=cuda",
        "--env.type=libero",
        f"--env.task={suite}",
        # No task_ids filter: evaluate every task in the suite.
        "--env.max_parallel_tasks=1",
        "--eval.batch_size=1",
        f"--eval.n_episodes={episodes_per_task}",
        "--seed=1000",
        f"--output_dir={output_dir}",
    ]

    # Save detailed output to a log instead of flooding the notebook.
    with log_path.open("w") as log:
        process = subprocess.Popen(
            command,
            env=env,
            stdin=subprocess.PIPE,
            stdout=log,
            stderr=subprocess.STDOUT,
            text=True,
        )
        try:
            process.communicate(input="N\n")
        except BaseException:
            process.terminate()
            try:
                process.wait(timeout=15)
            except subprocess.TimeoutExpired:
                process.kill()
                process.wait()
            raise

    if process.returncode != 0:
        tail = log_path.read_text(errors="replace").splitlines()[-40:]
        print("\n".join(tail))
        raise RuntimeError(f"{suite} failed. See {log_path}")

    results = json.loads((output_dir / "eval_info.json").read_text())
    metrics = results["overall"]
    summary[suite] = metrics

    (root_dir / "summary.json").write_text(
        json.dumps(summary, indent=2)
    )

    print(
        f"{suite}: {metrics['pc_success']:.1f}% success "
        f"over {metrics['n_episodes']} episodes",
        flush=True,
    )

total = sum(m["n_episodes"] for m in summary.values())
overall_success = sum(
    m["pc_success"] * m["n_episodes"]
    for m in summary.values()
) / total

print("\nAll evaluations complete!")
print(f"Total attempts: {total}")
print(f"Overall success: {overall_success:.1f}%")
print(f"Results and videos: {root_dir.resolve()}")
```

```python
import gc
import json
import math
import random
import shutil
import time
import warnings
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch

from lerobot.configs.policies import PreTrainedConfig
from lerobot.configs.types import FeatureType, PolicyFeature
from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy

PROJECT = Path("/home/hasith/Desktop/SmolVLA")
OLD_WORK = PROJECT / "outputs/finetuning/base16_to_libero_20261120_185029"

BASE_REVISION = "5e8d12a6e2975b0e5e5fce7c8caf47c371d257b6"
BASE_PATH = (
    Path.home() / ".cache/huggingface/hub"
    / "models--lerobot--smolvla_base/snapshots" / BASE_REVISION
)
assert (BASE_PATH / "model.safetensors").exists()
assert torch.cuda.is_available()

DATASET_ID = "lerobot/libero"
DATASET_REVISION = "a1aaacb7f6cd6ee5fb43120f673cebb0cfea7dd4"

STEPS = 10_000
BATCH_SIZE = 16
SAVE_EVERY = 5_000
PEAK_LR = 1e-4
END_LR = 2.5e-6
WARMUP = 1_000
SEED = 1000

split = json.loads((OLD_WORK / "episode_split.json").read_text())
assert not set(split["train"]) & set(split["validation"])

# Reuse the same validation sample selection.
indices_file = (
    OLD_WORK / "finetune_20k_bs4_20261121_113721"
    / "validation_indices.json"
)
val_indices = json.loads(indices_file.read_text())

RUN = PROJECT / "outputs/finetuning" / (
    f"base16_corrected_from_start_100k_{datetime.now():%Y%m%d_%H%M%S}"
)
RUN.mkdir(parents=True, exist_ok=False)

for name in (
    "policy", "combined", "optimizer", "scheduler", "trainable",
    "loss", "batch", "raw_batch", "iterator",
):
    globals().pop(name, None)

gc.collect()
torch.cuda.empty_cache()

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

cfg = PreTrainedConfig.from_pretrained(BASE_PATH)
assert isinstance(cfg, SmolVLAConfig)
assert cfg.num_vlm_layers == 16

# Adapt the interfaces to LIBERO; internal padded dimensions stay unchanged.
cfg.input_features = {
    "observation.images.camera1": PolicyFeature(
        type=FeatureType.VISUAL, shape=(3, 256, 256)
    ),
    "observation.images.camera2": PolicyFeature(
        type=FeatureType.VISUAL, shape=(3, 256, 256)
    ),
    "observation.state": PolicyFeature(
        type=FeatureType.STATE, shape=(8,)
    ),
}
cfg.output_features = {
    "action": PolicyFeature(type=FeatureType.ACTION, shape=(7,))
}
cfg.device = "cuda"
cfg.push_to_hub = False
cfg.freeze_vision_encoder = True
cfg.train_expert_only = True
cfg.train_state_proj = True
cfg.n_action_steps = 1

warnings.filterwarnings(
    "ignore",
    message=r"The video decoding and encoding capabilities of torchvision.*",
    category=UserWarning,
)

(RUN / "episode_split.json").write_text(json.dumps(split, indent=2))
(RUN / "validation_indices.json").write_text(json.dumps(val_indices))

print("Starting weights:", BASE_PATH)
print("Training / validation episodes:",
      len(split["train"]), "/", len(split["validation"]))
print("Free disk space:", round(shutil.disk_usage(RUN).free / 2**30, 1), "GiB")
print("Output:", RUN)
```

```python
from torch.utils.data import DataLoader, Subset
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

meta = LeRobotDatasetMetadata(
    DATASET_ID, revision=DATASET_REVISION
)
delta_timestamps = resolve_delta_timestamps(cfg, meta)

common = {
    "repo_id": DATASET_ID,
    "revision": DATASET_REVISION,
    "delta_timestamps": delta_timestamps,
    "video_backend": "pyav",
}

train_dataset = LeRobotDataset(
    **common, episodes=split["train"]
)
val_dataset = LeRobotDataset(
    **common, episodes=split["validation"]
)

assert set(train_dataset.episodes) == set(split["train"])
assert set(val_dataset.episodes) == set(split["validation"])
assert min(val_indices) >= 0
assert max(val_indices) < len(val_dataset)

# Read numerical columns only—no image decoding.
numeric = (
    train_dataset.hf_dataset
    .select_columns(["observation.state", "action"])
    .to_pandas()
)

train_stats = {}
for key in ["observation.state", "action"]:
    values = np.stack(numeric[key].to_numpy()).astype(np.float64)
    assert np.isfinite(values).all()

    train_stats[key] = {
        "mean": torch.tensor(values.mean(axis=0), dtype=torch.float32),
        "std": torch.tensor(values.std(axis=0), dtype=torch.float32),
        "min": torch.tensor(values.min(axis=0), dtype=torch.float32),
        "max": torch.tensor(values.max(axis=0), dtype=torch.float32),
    }

del numeric, values

(RUN / "training_statistics.json").write_text(json.dumps({
    key: {name: value.tolist() for name, value in stats.items()}
    for key, stats in train_stats.items()
}, indent=2))

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    drop_last=True,
    num_workers=0,
    generator=torch.Generator().manual_seed(SEED),
)
val_loader = DataLoader(
    Subset(val_dataset, val_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

print("Training samples:", len(train_dataset))
print("Fixed validation samples:", len(val_indices))
print("Approximate epochs:", round(STEPS / len(train_loader), 3))
```

```python
from lerobot.policies.factory import make_pre_post_processors

# Load original base weights, with the LIBERO interface configuration.
policy = SmolVLAPolicy.from_pretrained(
    BASE_PATH,
    config=cfg,
    strict=True,
).to("cuda")

preprocess, postprocess = make_pre_post_processors(
    policy.config,
    dataset_stats=train_stats,
)

RENAME_MAP = {
    "observation.images.image": "observation.images.camera1",
    "observation.images.image2": "observation.images.camera2",
}

# Configure the new processor's first step explicitly.
assert hasattr(preprocess.steps[0], "rename_map")
preprocess.steps[0].rename_map = RENAME_MAP

assert all(
    not p.requires_grad
    for p in policy.model.vlm_with_expert.vlm.parameters()
)
trainable = [p for p in policy.parameters() if p.requires_grad]

INITIAL = RUN / "checkpoints/000000/pretrained_model"
INITIAL.mkdir(parents=True)
policy.save_pretrained(INITIAL)
preprocess.save_pretrained(INITIAL)
postprocess.save_pretrained(INITIAL)

(RUN / "experiment.json").write_text(json.dumps({
    "model_id": "lerobot/smolvla_base",
    "model_revision": BASE_REVISION,
    "dataset_id": DATASET_ID,
    "dataset_revision": DATASET_REVISION,
    "steps": STEPS,
    "batch_size": BATCH_SIZE,
    "seed": SEED,
    "peak_lr": PEAK_LR,
    "end_lr": END_LR,
    "warmup": WARMUP,
    "objective": "valid timesteps and seven real action dimensions",
    "normalization": "training episodes only",
    "starting_checkpoint": str(BASE_PATH),
    "optimizer": "fresh AdamW",
}, indent=2))

print("Original base weights loaded.")
print("Trainable parameters:",
      round(sum(p.numel() for p in trainable) / 1e6, 2), "M")
print("Initial checkpoint saved:", INITIAL)
```

```python
def reduce_valid_losses(element_losses, padding_mask, action_dim):
    real = element_losses[..., :action_dim]
    valid = ~padding_mask.to(device=real.device, dtype=torch.bool)

    assert valid.shape == real.shape[:2]
    denominator = valid.sum() * action_dim
    if denominator.item() == 0:
        raise RuntimeError("No valid action targets in batch.")

    return (
        real * valid.unsqueeze(-1).to(real.dtype)
    ).sum() / denominator


def corrected_loss(batch):
    assert not policy.config.adapt_to_pi_aloha
    if "action_is_pad" not in batch:
        raise RuntimeError("Missing action_is_pad.")

    images, image_masks = policy.prepare_images(batch)
    state = policy.prepare_state(batch)
    actions = policy.prepare_action(batch)

    element_losses = policy.model.forward(
        images,
        image_masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        state,
        actions,
    )
    return reduce_valid_losses(
        element_losses,
        batch["action_is_pad"],
        policy.config.action_feature.shape[0],
    )


# Padded dimensions and timesteps must not influence the result.
test_losses = torch.full((1, 4, 32), 1000.0)
test_losses[:, :2, :7] = 1.0
test_mask = torch.tensor([[False, False, True, True]])

assert torch.allclose(
    reduce_valid_losses(test_losses, test_mask, 7),
    torch.tensor(1.0),
)

# Verify a real training batch, including backward computation.
raw_check = next(iter(train_loader))
processed_check = preprocess(dict(raw_check))

assert processed_check["action"].shape[1:] == (50, 7)
assert processed_check["action_is_pad"].shape[1] == 50
assert "observation.images.camera1" in processed_check
assert "observation.images.camera2" in processed_check

policy.train()
check_loss = corrected_loss(processed_check)
assert torch.isfinite(check_loss)
check_loss.backward()

assert any(p.grad is not None for p in trainable)
assert all(
    p.grad is None
    for p in policy.model.vlm_with_expert.vlm.parameters()
)

policy.zero_grad(set_to_none=True)
print("Masking, real-batch forward, and backward checks passed.")
print("Check loss:", check_loss.item())

del raw_check, processed_check, check_loss
```

```python
@torch.no_grad()
def validate_corrected():
    was_training = policy.training
    policy.eval()
    total_loss = 0.0
    total_targets = 0

    try:
        with torch.random.fork_rng(devices=[torch.cuda.current_device()]):
            torch.manual_seed(SEED)

            for raw in val_loader:
                batch = preprocess(dict(raw))
                loss = corrected_loss(batch)
                if not torch.isfinite(loss):
                    raise RuntimeError("Non-finite validation loss.")

                targets = int(
                    (~batch["action_is_pad"].bool()).sum()
                ) * 7
                total_loss += loss.item() * targets
                total_targets += targets
    finally:
        policy.train(was_training)

    return total_loss / total_targets


def save_training_checkpoint(step):
    destination = RUN / "checkpoints" / f"{step:06d}"
    model_dir = destination / "pretrained_model"
    model_dir.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(model_dir)
    preprocess.save_pretrained(model_dir)
    postprocess.save_pretrained(model_dir)

    torch.save({
        "step": step,
        "optimizer": optimizer.state_dict(),
        "torch_rng": torch.get_rng_state(),
        "cuda_rng": torch.cuda.get_rng_state_all(),
        "python_rng": random.getstate(),
        "numpy_rng": np.random.get_state(),
        "loader_generator_rng": train_loader.generator.get_state(),
        "schedule": {
            "steps": STEPS,
            "warmup": WARMUP,
            "peak_lr": PEAK_LR,
            "end_lr": END_LR,
        },
        "objective": "valid_timesteps_and_7_action_dimensions",
    }, destination / "training_state.pt")

    return model_dir
```

```python
from tqdm import tqdm

optimizer = torch.optim.AdamW(
    trainable,
    lr=PEAK_LR,
    betas=tuple(policy.config.optimizer_betas),
    eps=policy.config.optimizer_eps,
    weight_decay=policy.config.optimizer_weight_decay,
)

def learning_rate(step):
    if step <= WARMUP:
        return PEAK_LR * step / WARMUP

    fraction = (step - WARMUP) / (STEPS - WARMUP)
    return END_LR + 0.5 * (PEAK_LR - END_LR) * (
        1 + math.cos(math.pi * fraction)
    )

validation_rows = [{
    "step": 0,
    "corrected_validation_loss": validate_corrected(),
    "checkpoint": str(INITIAL),
}]
training_rows = []

pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv", index=False
)
print("Starting validation:",
      validation_rows[0]["corrected_validation_loss"], flush=True)

iterator = iter(train_loader)
policy.train()
started = time.monotonic()

for step in tqdm(range(1, STEPS + 1), desc="Fresh corrected fine-tuning"):
    lr = learning_rate(step)
    for group in optimizer.param_groups:
        group["lr"] = lr

    try:
        raw_batch = next(iterator)
    except StopIteration:
        iterator = iter(train_loader)
        raw_batch = next(iterator)

    batch = preprocess(dict(raw_batch))
    optimizer.zero_grad(set_to_none=True)

    loss = corrected_loss(batch)
    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite loss at step {step}")

    loss.backward()
    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable,
        policy.config.optimizer_grad_clip_norm,
        error_if_nonfinite=True,
    )
    optimizer.step()

    if step == 1 or step % 100 == 0:
        training_rows.append({
            "step": step,
            "loss": loss.item(),
            "grad_norm": float(grad_norm),
            "lr": lr,
            "elapsed_seconds": time.monotonic() - started,
        })
        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv", index=False
        )

    if step % SAVE_EVERY == 0:
        model_dir = save_training_checkpoint(step)
        score = validate_corrected()

        validation_rows.append({
            "step": step,
            "corrected_validation_loss": score,
            "checkpoint": str(model_dir),
        })
        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv", index=False
        )

        tqdm.write(
            f"Step {step}: validation={score:.6f}; saved {model_dir}"
        )

print("Training finished:", RUN)
display(pd.DataFrame(validation_rows))
```

```python
import gc
import json
import random
import time
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from torch.utils.data import DataLoader, Subset
from tqdm.auto import tqdm
from IPython.display import display

from lerobot.configs.policies import PreTrainedConfig
from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy
from lerobot.policies.factory import make_pre_post_processors
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

PROJECT = Path("/home/hasith/Desktop/SmolVLA")
SOURCE = (
    PROJECT / "outputs/finetuning"
    / "base16_corrected_from_start_100k_20261122_154723"
)
START = SOURCE / "checkpoints/010000"
MODEL = START / "pretrained_model"

TARGET_STEP = 50000
BATCH_SIZE = 16
SAVE_EVERY = 5000
SEED = 1000

assert torch.cuda.is_available()
assert (MODEL / "model.safetensors").exists()

# Load our own trusted training-state file.
state = torch.load(
    START / "training_state.pt",
    map_location="cpu",
    weights_only=False,
)
START_STEP = int(state["step"])
assert START_STEP == 10000

experiment = json.loads((SOURCE / "experiment.json").read_text())
assert experiment["batch_size"] == BATCH_SIZE

RUN = SOURCE / f"continue_to_50k_{datetime.now():%Y%m%d_%H%M%S}"
RUN.mkdir(parents=True, exist_ok=False)

config = PreTrainedConfig.from_pretrained(str(MODEL))
assert isinstance(config, SmolVLAConfig)
assert config.num_vlm_layers == 16
config.device = "cuda"

policy = SmolVLAPolicy.from_pretrained(
    str(MODEL), config=config, strict=True
).to("cuda")

# Reuse the saved normalization statistics and camera mapping.
preprocess, postprocess = make_pre_post_processors(
    policy.config,
    pretrained_path=str(MODEL),
)

trainable = [p for p in policy.parameters() if p.requires_grad]
assert trainable

optimizer = torch.optim.AdamW(
    trainable,
    lr=2.5e-6,
    betas=(0.9, 0.95),
    eps=1e-8,
    weight_decay=1e-10,
)
optimizer.load_state_dict(state["optimizer"])

# Explicitly preserve the learning rate reached at step 10,000.
CONTINUE_LR = float(optimizer.param_groups[0]["lr"])
assert all(
    group["lr"] == CONTINUE_LR
    for group in optimizer.param_groups
)

# Keep copies of the original split and validation sample selection.
for filename in ("episode_split.json", "validation_indices.json"):
    (RUN / filename).write_text((SOURCE / filename).read_text())

(RUN / "experiment.json").write_text(json.dumps({
    **experiment,
    "starting_checkpoint": str(MODEL),
    "start_step": START_STEP,
    "target_step": TARGET_STEP,
    "additional_steps": TARGET_STEP - START_STEP,
    "batch_size": BATCH_SIZE,
    "save_every": SAVE_EVERY,
    "learning_rate": CONTINUE_LR,
    "schedule": "constant at saved optimizer learning rate",
    "optimizer": "restored AdamW state",
}, indent=2))

print("Loaded checkpoint:", MODEL)
print("Training:", START_STEP, "→", TARGET_STEP)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", CONTINUE_LR)
print("Output:", RUN)
```

```python
split = json.loads((SOURCE / "episode_split.json").read_text())
train_episodes = split["train"]
validation_episodes = split["validation"]

assert not set(train_episodes) & set(validation_episodes)

metadata = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)
delta_timestamps = resolve_delta_timestamps(policy.config, metadata)

train_dataset = LeRobotDataset(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
    episodes=train_episodes,
    delta_timestamps=delta_timestamps,
    video_backend="pyav",
)
val_dataset = LeRobotDataset(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
    episodes=validation_episodes,
    delta_timestamps=delta_timestamps,
    video_backend="pyav",
)

validation_indices = json.loads(
    (SOURCE / "validation_indices.json").read_text()
)
assert isinstance(validation_indices, list)
assert max(validation_indices) < len(val_dataset)

loader_generator = torch.Generator()
loader_generator.set_state(state["loader_generator_rng"])

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    generator=loader_generator,
    num_workers=0,
    drop_last=True,
)
val_loader = DataLoader(
    Subset(val_dataset, validation_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

def corrected_loss(batch):
    images, image_masks = policy.prepare_images(batch)
    robot_state = policy.prepare_state(batch)
    actions = policy.prepare_action(batch)

    element_losses = policy.model.forward(
        images,
        image_masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        robot_state,
        actions,
    )

    action_dim = policy.config.action_feature.shape[0]
    real_losses = element_losses[..., :action_dim]
    valid = ~batch["action_is_pad"].to(
        device=real_losses.device, dtype=torch.bool
    )

    assert valid.shape == real_losses.shape[:2]
    denominator = valid.sum() * action_dim
    assert denominator.item() > 0

    return (
        real_losses * valid.unsqueeze(-1).to(real_losses.dtype)
    ).sum() / denominator


@torch.no_grad()
def validation_loss():
    was_training = policy.training
    policy.eval()
    numerator = 0.0
    denominator = 0

    try:
        with torch.random.fork_rng(
            devices=[torch.cuda.current_device()]
        ):
            torch.manual_seed(SEED)

            for raw in val_loader:
                processed = preprocess(dict(raw))
                value = corrected_loss(processed)
                if not torch.isfinite(value):
                    raise RuntimeError("Non-finite validation loss.")

                count = (
                    int((~processed["action_is_pad"].bool()).sum())
                    * policy.config.action_feature.shape[0]
                )
                numerator += value.item() * count
                denominator += count
    finally:
        policy.train(was_training)

    return numerator / denominator


# Restore random states after constructing/loading everything.
torch.set_rng_state(state["torch_rng"])
torch.cuda.set_rng_state_all(state["cuda_rng"])
random.setstate(state["python_rng"])
np.random.set_state(state["numpy_rng"])

print("Training episodes:", len(train_episodes))
print("Validation episodes:", len(validation_episodes))
print("Validation samples:", len(validation_indices))
print("Starting validation loss:", validation_loss())
```

```python
def save_checkpoint(step):
    destination = RUN / "checkpoints" / f"{step:06d}"
    model_dir = destination / "pretrained_model"
    model_dir.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(model_dir)
    preprocess.save_pretrained(model_dir)
    postprocess.save_pretrained(model_dir)

    torch.save({
        "step": step,
        "optimizer": optimizer.state_dict(),
        "torch_rng": torch.get_rng_state(),
        "cuda_rng": torch.cuda.get_rng_state_all(),
        "python_rng": random.getstate(),
        "numpy_rng": np.random.get_state(),
        "loader_generator_rng": loader_generator.get_state(),
        "schedule": {
            "type": "constant",
            "lr": CONTINUE_LR,
            "target_step": TARGET_STEP,
        },
        "objective": "valid_timesteps_and_7_action_dimensions",
    }, destination / "training_state.pt")

    return model_dir


training_rows = []
validation_rows = [{
    "step": START_STEP,
    "corrected_validation_loss": validation_loss(),
    "checkpoint": str(MODEL),
}]
pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv", index=False
)

iterator = iter(train_loader)
started = time.monotonic()
policy.train()

for step in tqdm(
    range(START_STEP + 1, TARGET_STEP + 1),
    total=TARGET_STEP - START_STEP,
    desc="Corrected fine-tuning: 10k → 50k",
):
    try:
        raw = next(iterator)
    except StopIteration:
        iterator = iter(train_loader)
        raw = next(iterator)

    batch = preprocess(dict(raw))
    optimizer.zero_grad(set_to_none=True)

    loss = corrected_loss(batch)
    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite training loss at step {step}")

    loss.backward()
    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable, max_norm=10.0, error_if_nonfinite=True
    )
    optimizer.step()

    if step % 100 == 0:
        training_rows.append({
            "step": step,
            "loss": loss.item(),
            "grad_norm": float(grad_norm),
            "lr": optimizer.param_groups[0]["lr"],
            "elapsed_seconds": time.monotonic() - started,
        })
        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv", index=False
        )

    if step % SAVE_EVERY == 0:
        # Save before validation so training progress is protected.
        model_dir = save_checkpoint(step)
        score = validation_loss()

        validation_rows.append({
            "step": step,
            "corrected_validation_loss": score,
            "checkpoint": str(model_dir),
        })
        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv", index=False
        )

        tqdm.write(
            f"Step {step}: validation={score:.6f}; "
            f"saved {model_dir}"
        )

print("Finished at total step:", TARGET_STEP)
print("Results:", RUN)
display(pd.DataFrame(validation_rows))
```

```python
import json
import time
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from torch.utils.data import DataLoader, Subset, WeightedRandomSampler
from tqdm.auto import tqdm
from IPython.display import display

from lerobot.configs.policies import PreTrainedConfig
from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy
from lerobot.policies.factory import make_pre_post_processors
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

PROJECT = Path("/home/hasith/Desktop/SmolVLA")
SOURCE = (
    PROJECT / "outputs/finetuning"
    / "base16_corrected_from_start_100k_20261122_154723"
)
START = (
    SOURCE / "continue_to_50k_20261123_003642"
    / "checkpoints/050000"
)
MODEL = START / "pretrained_model"

BATCH_SIZE = 16
ADDITIONAL_STEPS = 5000
SAVE_EVERY = 1000
SEED = 1000

assert torch.cuda.is_available()
assert (MODEL / "model.safetensors").exists()

experiment = json.loads((SOURCE / "experiment.json").read_text())
split = json.loads((SOURCE / "episode_split.json").read_text())

# Simulator task IDs differ from dataset task IDs.
SIM_TO_DATASET = {
    0: 5, 1: 7, 2: 3, 3: 8, 4: 0,
    5: 9, 6: 1, 7: 4, 8: 6, 9: 2,
}
WEAK_SIM_TASKS = [0, 4, 6, 7]
WEAK_DATASET_TASKS = {SIM_TO_DATASET[t] for t in WEAK_SIM_TASKS}
LIBERO10_DATASET_TASKS = set(SIM_TO_DATASET.values())

RUN = (
    SOURCE
    / f"weak_long_tasks_pilot_{datetime.now():%Y%m%d_%H%M%S}"
)
RUN.mkdir(parents=True, exist_ok=False)

config = PreTrainedConfig.from_pretrained(str(MODEL))
assert isinstance(config, SmolVLAConfig)
config.device = "cuda"

policy = SmolVLAPolicy.from_pretrained(
    str(MODEL), config=config, strict=True
).to("cuda")

preprocess, postprocess = make_pre_post_processors(
    policy.config,
    pretrained_path=str(MODEL),
)

trainable = [p for p in policy.parameters() if p.requires_grad]

optimizer = torch.optim.AdamW(
    trainable,
    lr=2.5e-6,
    betas=(0.9, 0.95),
    eps=1e-8,
    weight_decay=1e-10,
)

# This is our own trusted checkpoint.
saved_state = torch.load(
    START / "training_state.pt",
    map_location="cpu",
    weights_only=False,
)
assert saved_state["step"] == 50000
optimizer.load_state_dict(saved_state["optimizer"])

for group in optimizer.param_groups:
    group["lr"] = 2.5e-6

print("Starting from:", MODEL)
print("Total training steps: 50,000 → 55,000")
print("Output:", RUN)
```

```python
metadata = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)
delta_timestamps = resolve_delta_timestamps(policy.config, metadata)

# Read identifiers without decoding video frames.
index_dataset = LeRobotDataset(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
    video_backend="pyav",
)
episode_tasks = (
    index_dataset.hf_dataset
    .select_columns(["episode_index", "task_index"])
    .to_pandas()
    .drop_duplicates()
)

assert (
    episode_tasks.groupby("episode_index")["task_index"].nunique().max()
    == 1
)

long_episode_ids = set(
    episode_tasks.loc[
        episode_tasks["task_index"].isin(LIBERO10_DATASET_TASKS),
        "episode_index",
    ].astype(int)
)

train_ids = sorted(long_episode_ids & set(split["train"]))
val_ids = sorted(long_episode_ids & set(split["validation"]))
assert not set(train_ids) & set(val_ids)

del index_dataset

train_dataset = LeRobotDataset(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
    episodes=train_ids,
    delta_timestamps=delta_timestamps,
    video_backend="pyav",
)
val_dataset = LeRobotDataset(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
    episodes=val_ids,
    delta_timestamps=delta_timestamps,
    video_backend="pyav",
)

frame_tasks = np.asarray(
    train_dataset.hf_dataset["task_index"], dtype=np.int64
)
assert set(np.unique(frame_tasks)) == LIBERO10_DATASET_TASKS

weights = np.zeros(len(frame_tasks), dtype=np.float64)

# Each weak task gets 12.5% probability; together, 50%.
# Each remaining task gets approximately 8.33%; together, 50%.
for task_id in LIBERO10_DATASET_TASKS:
    mask = frame_tasks == task_id
    task_probability = (
        0.5 / len(WEAK_DATASET_TASKS)
        if task_id in WEAK_DATASET_TASKS
        else 0.5 / (
            len(LIBERO10_DATASET_TASKS) - len(WEAK_DATASET_TASKS)
        )
    )
    weights[mask] = task_probability / int(mask.sum())

sampler_generator = torch.Generator().manual_seed(SEED)
sampler = WeightedRandomSampler(
    torch.from_numpy(weights),
    num_samples=ADDITIONAL_STEPS * BATCH_SIZE,
    replacement=True,
    generator=sampler_generator,
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=0,
    drop_last=True,
)

# Fixed validation samples from the four weak tasks only.
val_frame_tasks = np.asarray(
    val_dataset.hf_dataset["task_index"], dtype=np.int64
)
rng = np.random.default_rng(SEED)
validation_indices = []

for task_id in sorted(WEAK_DATASET_TASKS):
    indices = np.flatnonzero(val_frame_tasks == task_id)
    assert len(indices) > 0
    selected = rng.choice(
        indices, size=min(256, len(indices)), replace=False
    )
    validation_indices.extend(selected.tolist())

val_loader = DataLoader(
    Subset(val_dataset, validation_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
)

(RUN / "episode_split.json").write_text(json.dumps({
    "train": train_ids,
    "validation": val_ids,
}, indent=2))
(RUN / "validation_indices.json").write_text(
    json.dumps(validation_indices, indent=2)
)
(RUN / "experiment.json").write_text(json.dumps({
    "starting_checkpoint": str(MODEL),
    "dataset_id": experiment["dataset_id"],
    "dataset_revision": experiment["dataset_revision"],
    "start_step": 50000,
    "additional_steps": ADDITIONAL_STEPS,
    "batch_size": BATCH_SIZE,
    "lr": 2.5e-6,
    "weak_simulator_tasks": WEAK_SIM_TASKS,
    "weak_dataset_tasks": sorted(WEAK_DATASET_TASKS),
    "weak_task_sampling_probability": 0.5,
    "sampling": "replacement; balanced within each task group",
    "objective": "valid timesteps and seven real action dimensions",
    "validation": "fixed samples from reserved weak-task episodes",
}, indent=2))

print("LIBERO_10 training demonstrations:", len(train_ids))
print("Reserved LIBERO_10 validation demonstrations:", len(val_ids))
print("Weak-task validation samples:", len(validation_indices))
print("Training batches:", len(train_loader))
```

```python
def corrected_loss(batch):
    images, image_masks = policy.prepare_images(batch)
    robot_state = policy.prepare_state(batch)
    actions = policy.prepare_action(batch)

    element_losses = policy.model.forward(
        images,
        image_masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        robot_state,
        actions,
    )

    action_dim = policy.config.action_feature.shape[0]
    assert action_dim == 7

    real_losses = element_losses[..., :action_dim]
    valid = ~batch["action_is_pad"].to(
        device=real_losses.device, dtype=torch.bool
    )
    assert valid.shape == real_losses.shape[:2]

    denominator = valid.sum() * action_dim
    assert denominator.item() > 0

    return (
        real_losses * valid.unsqueeze(-1).to(real_losses.dtype)
    ).sum() / denominator


@torch.no_grad()
def weak_task_validation():
    was_training = policy.training
    policy.eval()
    numerator = 0.0
    denominator = 0

    try:
        with torch.random.fork_rng(
            devices=[torch.cuda.current_device()]
        ):
            torch.manual_seed(SEED)

            for raw in val_loader:
                batch = preprocess(dict(raw))
                value = corrected_loss(batch)

                if not torch.isfinite(value):
                    raise RuntimeError("Non-finite validation loss.")

                count = int(
                    (~batch["action_is_pad"].bool()).sum()
                ) * 7

                numerator += value.item() * count
                denominator += count
    finally:
        policy.train(was_training)

    return numerator / denominator


def save_pilot(step):
    destination = RUN / "checkpoints" / f"{step:06d}"
    model_dir = destination / "pretrained_model"
    model_dir.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(model_dir)
    preprocess.save_pretrained(model_dir)
    postprocess.save_pretrained(model_dir)

    torch.save({
        "step": step,
        "optimizer": optimizer.state_dict(),
        "torch_rng": torch.get_rng_state(),
        "cuda_rng": torch.cuda.get_rng_state_all(),
        "sampler_generator_rng": sampler_generator.get_state(),
        "objective": "valid_timesteps_and_7_action_dimensions",
        "lr": 2.5e-6,
    }, destination / "training_state.pt")

    return model_dir


torch.manual_seed(SEED)

baseline = weak_task_validation()
validation_rows = [{
    "step": 50000,
    "weak_task_validation_loss": baseline,
    "checkpoint": str(MODEL),
}]
training_rows = []

pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv", index=False
)

print("Starting weak-task validation loss:", baseline)
started = time.monotonic()
policy.train()

# Run this training cell once. Re-running starts another sampler pass.
for local_step, raw in enumerate(
    tqdm(train_loader, desc="Weak-task pilot: 50k → 55k"),
    start=1,
):
    global_step = 50000 + local_step

    batch = preprocess(dict(raw))
    optimizer.zero_grad(set_to_none=True)

    loss = corrected_loss(batch)
    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite loss at step {global_step}")

    loss.backward()
    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable,
        max_norm=10.0,
        error_if_nonfinite=True,
    )
    optimizer.step()

    if local_step % 100 == 0:
        training_rows.append({
            "step": global_step,
            "loss": loss.item(),
            "grad_norm": float(grad_norm),
            "elapsed_seconds": time.monotonic() - started,
        })
        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv", index=False
        )

    if local_step % SAVE_EVERY == 0:
        model_dir = save_pilot(global_step)
        score = weak_task_validation()

        validation_rows.append({
            "step": global_step,
            "weak_task_validation_loss": score,
            "checkpoint": str(model_dir),
        })
        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv", index=False
        )

        tqdm.write(
            f"Step {global_step}: "
            f"weak-task validation={score:.6f}; saved {model_dir}"
        )

print("Pilot finished at total step 55,000.")
print("Results:", RUN)
display(pd.DataFrame(validation_rows))
```

```python
index_file = EVAL_ROOT / "completed_suite_paths.json"
completed_paths = (
    json.loads(index_file.read_text())
    if index_file.exists()
    else {}
)

summary = {}
task_rows = []

for suite in SUITES:
    saved = completed_paths.get(suite)
    result_file = Path(saved) if saved else None

    if result_file is None or not result_file.exists():
        candidates = [
            EVAL_ROOT / suite / "eval_info.json",
            *sorted(EVAL_ROOT.glob(
                f"{suite}_retry_*/eval_info.json"
            )),
        ]
        completed = [p for p in candidates if p.exists()]
        result_file = completed[-1] if completed else None

    if result_file is None:
        output_dir = EVAL_ROOT / suite

        if output_dir.exists():
            output_dir = EVAL_ROOT / (
                f"{suite}_retry_{datetime.now():%Y%m%d_%H%M%S_%f}"
            )

        log_path = EVAL_ROOT / f"{output_dir.name}_evaluation.log"

        command = [
            sys.executable, "-m", "lerobot.scripts.lerobot_eval",
            f"--policy.path={CHECKPOINT}",
            "--policy.device=cuda",
            "--policy.n_action_steps=1",
            "--env.type=libero",
            f"--env.task={suite}",
            "--env.max_parallel_tasks=1",
            "--eval.batch_size=1",
            "--eval.n_episodes=10",
            "--seed=1000",
            f"--rename_map={json.dumps(RENAME_MAP)}",
            f"--output_dir={output_dir}",
        ]

        print(f"\nStarting {suite}: 100 attempts", flush=True)
        print("Progress log:", log_path, flush=True)

        with log_path.open("w") as log:
            process = subprocess.Popen(
                command,
                cwd=PROJECT,
                env=env,
                stdin=subprocess.PIPE,
                stdout=log,
                stderr=subprocess.STDOUT,
                text=True,
            )

            try:
                process.communicate(input="N\n")
            except BaseException:
                process.terminate()
                try:
                    process.wait(timeout=15)
                except subprocess.TimeoutExpired:
                    process.kill()
                    process.wait()
                raise

        if process.returncode != 0:
            lines = log_path.read_text(
                errors="replace"
            ).splitlines()

            first_traceback = next(
                (i for i, line in enumerate(lines)
                 if line.startswith("Traceback")),
                max(0, len(lines) - 80),
            )

            print("\n".join(
                lines[first_traceback:first_traceback + 80]
            ))
            raise RuntimeError(f"Evaluation failed: {log_path}")

        result_file = output_dir / "eval_info.json"

    else:
        print(f"Reusing completed suite: {suite}", flush=True)

    result = json.loads(result_file.read_text())
    metrics = result["overall"]
    assert metrics["n_episodes"] == 100, metrics

    summary[suite] = metrics
    completed_paths[suite] = str(result_file)

    index_file.write_text(json.dumps(completed_paths, indent=2))
    (EVAL_ROOT / "summary.json").write_text(
        json.dumps(summary, indent=2)
    )

    # Record each task separately so weak-task changes are visible.
    for task in result["per_task"]:
        successes = task["metrics"]["successes"]

        task_rows.append({
            "suite": task["task_group"],
            "task_id": int(task["task_id"]),
            "episodes": len(successes),
            "successes": sum(bool(x) for x in successes),
            "success_percent": (
                100 * sum(bool(x) for x in successes)
                / len(successes)
            ),
        })

    pd.DataFrame(task_rows).to_csv(
        EVAL_ROOT / "per_task_results.csv", index=False
    )

    print(
        f"{suite}: 50k={baseline_50k[suite]:.1f}% "
        f"→ 55k={metrics['pc_success']:.1f}%",
        flush=True,
    )

    # The simulator process has exited, releasing its GPU allocations.

rows = [{
    "suite": suite,
    "episodes": summary[suite]["n_episodes"],
    "50k_percent": baseline_50k[suite],
    "55k_percent": summary[suite]["pc_success"],
    "change_points": (
        summary[suite]["pc_success"] - baseline_50k[suite]
    ),
} for suite in SUITES]

table = pd.DataFrame(rows)
total = int(table["episodes"].sum())

overall_55k = float(
    (table["55k_percent"] * table["episodes"]).sum() / total
)

comparison = pd.concat([
    table,
    pd.DataFrame([{
        "suite": "OVERALL",
        "episodes": total,
        "50k_percent": 56.5,
        "55k_percent": overall_55k,
        "change_points": overall_55k - 56.5,
    }]),
], ignore_index=True)

comparison.to_csv(EVAL_ROOT / "comparison.csv", index=False)

print("\nAll-suite comparison:")
display(comparison.round(2))

per_task = pd.DataFrame(task_rows)
print("\nLIBERO_10 task results:")
display(
    per_task.loc[per_task["suite"] == "libero_10"]
    .sort_values("task_id")
    .round(2)
)

print("Total attempts:", total)
print("Results and videos:", EVAL_ROOT)
```

```python
import gc
import json
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from torch.utils.data import DataLoader, Subset, WeightedRandomSampler
from tqdm.auto import tqdm
from IPython.display import display

from lerobot.configs.policies import PreTrainedConfig
from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy
from lerobot.policies.factory import make_pre_post_processors
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

PROJECT = Path("/home/hasith/Desktop/SmolVLA")
SOURCE = (
    PROJECT / "outputs/finetuning"
    / "base16_corrected_from_start_100k_20261122_154723"
)
START = (
    SOURCE / "continue_to_50k_20261123_003642"
    / "checkpoints/050000"
)
MODEL = START / "pretrained_model"

START_STEP = 50000
EXTRA_STEPS = 5000
BATCH_SIZE = 16
SAVE_EVERY = 1000
LR = 2.5e-6
SEED = 1000

# Experimental sampling settings.
EVENT_RADIUS = 5
EVENT_MIX = 0.50

assert torch.cuda.is_available()
assert (MODEL / "model.safetensors").exists()
assert (START / "training_state.pt").exists()

experiment = json.loads((SOURCE / "experiment.json").read_text())
split = json.loads((SOURCE / "episode_split.json").read_text())
validation_indices = json.loads(
    (SOURCE / "validation_indices.json").read_text()
)

assert not set(split["train"]) & set(split["validation"])

config = PreTrainedConfig.from_pretrained(str(MODEL))
assert isinstance(config, SmolVLAConfig)

PAIR_ROOT = SOURCE / (
    f"stage_sampling_experiment_{datetime.now():%Y%m%d_%H%M%S}"
)
PAIR_ROOT.mkdir(parents=True, exist_ok=False)

(PAIR_ROOT / "settings.json").write_text(json.dumps({
    "starting_checkpoint": str(MODEL),
    "start_step": START_STEP,
    "additional_steps_per_branch": EXTRA_STEPS,
    "batch_size": BATCH_SIZE,
    "lr": LR,
    "seed": SEED,
    "event_radius_frames": EVENT_RADIUS,
    "event_mix": EVENT_MIX,
    "event_definition": "gripper command changes within training episodes",
    "task_proportions": "preserved from uniform training-frame sampling",
    "optimizer": "both branches restore the same 50k AdamW state",
}, indent=2))

print("Both branches start from:", MODEL)
print("Experiment folder:", PAIR_ROOT)
```

```python
metadata = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)
delta_timestamps = resolve_delta_timestamps(config, metadata)

dataset_kwargs = {
    "repo_id": experiment["dataset_id"],
    "revision": experiment["dataset_revision"],
    "delta_timestamps": delta_timestamps,
    "video_backend": "pyav",
}

train_dataset = LeRobotDataset(
    episodes=split["train"], **dataset_kwargs
)
val_dataset = LeRobotDataset(
    episodes=split["validation"], **dataset_kwargs
)

# Read numeric columns without decoding camera frames.
table = (
    train_dataset.hf_dataset
    .select_columns([
        "episode_index", "frame_index", "task_index", "action"
    ])
    .to_pandas()
    .reset_index(drop=True)
)

assert len(table) == len(train_dataset)
assert set(table["episode_index"].astype(int)) == set(split["train"])

actions = np.stack(table["action"].to_numpy())
gripper = actions[:, -1]
assert actions.shape[1] == 7
assert np.all(np.isclose(np.abs(gripper), 1.0))

event_mask = np.zeros(len(table), dtype=bool)
event_counts = {}

# Process each episode separately, preventing boundary-crossing events.
for episode_id, group in table.groupby("episode_index", sort=False):
    ordered = group.sort_values("frame_index")
    indices = ordered.index.to_numpy()
    commands = gripper[indices]
    frames = ordered["frame_index"].to_numpy()

    changes = np.flatnonzero(
        np.abs(np.diff(commands)) > 1.0
    ) + 1

    event_counts[int(episode_id)] = len(changes)

    for position in changes:
        nearby = (
            np.abs(frames - frames[position]) <= EVENT_RADIUS
        )
        event_mask[indices[nearby]] = True

assert event_mask.any(), "No gripper-change windows were found."

task_ids = table["task_index"].to_numpy(dtype=np.int64)
n_samples = len(table)

ordinary_weights = np.full(n_samples, 1.0 / n_samples)
stage_weights = np.zeros(n_samples, dtype=np.float64)
sampling_rows = []

for task_id in sorted(np.unique(task_ids)):
    mask = task_ids == task_id
    event_indices = mask & event_mask

    count = int(mask.sum())
    event_count = int(event_indices.sum())

    # Same task probability as uniform frame sampling.
    task_probability = count / n_samples

    if event_count:
        stage_weights[mask] = (
            task_probability * (1 - EVENT_MIX) / count
        )
        stage_weights[event_indices] += (
            task_probability * EVENT_MIX / event_count
        )
    else:
        stage_weights[mask] = task_probability / count

    assert np.isclose(
        stage_weights[mask].sum(),
        ordinary_weights[mask].sum(),
    )

    sampling_rows.append({
        "dataset_task_id": int(task_id),
        "training_frames": count,
        "event_window_frames": event_count,
        "task_probability": task_probability,
        "ordinary_event_fraction": event_count / count,
        "stage_event_fraction": (
            stage_weights[event_indices].sum() / task_probability
        ),
    })

assert np.isclose(ordinary_weights.sum(), 1.0)
assert np.isclose(stage_weights.sum(), 1.0)

SAMPLING_WEIGHTS = {
    "control": ordinary_weights,
    "stage_balanced": stage_weights,
}

sampling_report = pd.DataFrame(sampling_rows)
sampling_report.to_csv(
    PAIR_ROOT / "sampling_report.csv", index=False
)

# Fixed validation sample selection, shared by both branches.
assert max(validation_indices) < len(val_dataset)
validation_subset = Subset(val_dataset, validation_indices)

print("Training demonstrations:", len(split["train"]))
print("Reserved validation demonstrations:", len(split["validation"]))
print("Training frames:", n_samples)
print("Detected gripper transitions:", sum(event_counts.values()))
print(
    "Ordinary sampling: expected event-window share:",
    f"{ordinary_weights[event_mask].sum():.1%}",
)
print(
    "Stage-balanced sampling: expected event-window share:",
    f"{stage_weights[event_mask].sum():.1%}",
)

display(sampling_report.round(4))

# These numeric arrays are no longer needed.
del table, actions, gripper
```

```python
def train_branch(branch_name):
    branch_dir = PAIR_ROOT / branch_name
    branch_dir.mkdir(parents=True, exist_ok=False)

    cfg = PreTrainedConfig.from_pretrained(str(MODEL))
    cfg.device = "cuda"

    policy = SmolVLAPolicy.from_pretrained(
        str(MODEL), config=cfg, strict=True
    ).to("cuda")

    preprocess, postprocess = make_pre_post_processors(
        policy.config,
        pretrained_path=str(MODEL),
    )

    assert all(
        not p.requires_grad
        for p in policy.model.vlm_with_expert.vlm.parameters()
    )

    trainable = [
        p for p in policy.parameters() if p.requires_grad
    ]

    optimizer = torch.optim.AdamW(
        trainable,
        lr=LR,
        betas=(0.9, 0.95),
        eps=1e-8,
        weight_decay=1e-10,
    )

    state = torch.load(
        START / "training_state.pt",
        map_location="cpu",
        weights_only=False,
    )
    assert state["step"] == START_STEP
    optimizer.load_state_dict(state["optimizer"])

    for group in optimizer.param_groups:
        group["lr"] = LR

    # Separate generators prevent sampler draws from consuming
    # the flow-matching noise generator.
    sampler_generator = torch.Generator().manual_seed(SEED)
    loader_generator = torch.Generator().manual_seed(SEED + 1)
    val_generator = torch.Generator().manual_seed(SEED + 2)

    sampler = WeightedRandomSampler(
        torch.from_numpy(SAMPLING_WEIGHTS[branch_name]),
        num_samples=EXTRA_STEPS * BATCH_SIZE,
        replacement=True,
        generator=sampler_generator,
    )

    loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        sampler=sampler,
        generator=loader_generator,
        num_workers=0,
        drop_last=True,
    )

    val_loader = DataLoader(
        validation_subset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        generator=val_generator,
        num_workers=0,
    )

    assert len(loader) == EXTRA_STEPS

    # Same starting flow-matching random state for both branches.
    torch.set_rng_state(state["torch_rng"])
    torch.cuda.set_rng_state_all(state["cuda_rng"])
    del state

    for filename in ("episode_split.json", "validation_indices.json"):
        (branch_dir / filename).write_text(
            (SOURCE / filename).read_text()
        )

    (branch_dir / "experiment.json").write_text(json.dumps({
        **experiment,
        "branch": branch_name,
        "starting_checkpoint": str(MODEL),
        "start_step": START_STEP,
        "target_step": START_STEP + EXTRA_STEPS,
        "additional_steps": EXTRA_STEPS,
        "batch_size": BATCH_SIZE,
        "lr": LR,
        "event_radius_frames": EVENT_RADIUS,
        "event_mix": EVENT_MIX if branch_name == "stage_balanced" else 0,
        "sampling": "replacement; original task probabilities preserved",
    }, indent=2))

    def loss_fn(batch):
        images, masks = policy.prepare_images(batch)

        losses = policy.model.forward(
            images,
            masks,
            batch["observation.language.tokens"],
            batch["observation.language.attention_mask"],
            policy.prepare_state(batch),
            policy.prepare_action(batch),
        )

        action_dim = policy.config.action_feature.shape[0]
        assert action_dim == 7

        real = losses[..., :action_dim]
        valid = ~batch["action_is_pad"].to(
            device=real.device, dtype=torch.bool
        )

        assert valid.shape == real.shape[:2]
        denominator = valid.sum() * action_dim
        assert denominator.item() > 0

        return (
            real * valid.unsqueeze(-1).to(real.dtype)
        ).sum() / denominator

    @torch.no_grad()
    def validate():
        was_training = policy.training
        policy.eval()
        numerator, denominator = 0.0, 0

        try:
            with torch.random.fork_rng(
                devices=[torch.cuda.current_device()]
            ):
                torch.manual_seed(SEED)

                for raw in val_loader:
                    batch = preprocess(dict(raw))
                    value = loss_fn(batch)

                    if not torch.isfinite(value):
                        raise RuntimeError("Non-finite validation loss.")

                    count = int(
                        (~batch["action_is_pad"].bool()).sum()
                    ) * 7

                    numerator += value.item() * count
                    denominator += count
        finally:
            policy.train(was_training)

        return numerator / denominator

    def save(step):
        folder = branch_dir / "checkpoints" / f"{step:06d}"
        destination = folder / "pretrained_model"
        destination.mkdir(parents=True, exist_ok=False)

        policy.save_pretrained(destination)
        preprocess.save_pretrained(destination)
        postprocess.save_pretrained(destination)

        torch.save({
            "step": step,
            "optimizer": optimizer.state_dict(),
            "torch_rng": torch.get_rng_state(),
            "cuda_rng": torch.cuda.get_rng_state_all(),
            "sampler_generator_rng": sampler_generator.get_state(),
            "loader_generator_rng": loader_generator.get_state(),
            "branch": branch_name,
            "lr": LR,
            "objective": "valid_timesteps_and_7_action_dimensions",
        }, folder / "training_state.pt")

        return destination

    baseline = validate()
    validation_rows = [{
        "step": START_STEP,
        "validation_loss": baseline,
        "checkpoint": str(MODEL),
    }]
    training_rows = []

    pd.DataFrame(validation_rows).to_csv(
        branch_dir / "validation.csv", index=False
    )

    print(
        f"\n{branch_name}: starting validation={baseline:.6f}",
        flush=True,
    )

    policy.train()

    for local_step, raw in enumerate(
        tqdm(loader, desc=f"{branch_name}: 50k → 55k"),
        start=1,
    ):
        step = START_STEP + local_step
        batch = preprocess(dict(raw))

        optimizer.zero_grad(set_to_none=True)
        loss = loss_fn(batch)

        if not torch.isfinite(loss):
            raise RuntimeError(f"Non-finite loss at step {step}")

        loss.backward()
        grad_norm = torch.nn.utils.clip_grad_norm_(
            trainable, 10.0, error_if_nonfinite=True
        )
        optimizer.step()

        if local_step % 100 == 0:
            training_rows.append({
                "step": step,
                "loss": loss.item(),
                "grad_norm": float(grad_norm),
            })
            pd.DataFrame(training_rows).to_csv(
                branch_dir / "training.csv", index=False
            )

        if local_step % SAVE_EVERY == 0:
            destination = save(step)
            score = validate()

            validation_rows.append({
                "step": step,
                "validation_loss": score,
                "checkpoint": str(destination),
            })
            pd.DataFrame(validation_rows).to_csv(
                branch_dir / "validation.csv", index=False
            )

            tqdm.write(
                f"{branch_name}, step {step}: "
                f"validation={score:.6f}"
            )

    final_checkpoint = (
        branch_dir / "checkpoints/055000/pretrained_model"
    )
    print("Finished:", final_checkpoint, flush=True)

    # Return only CPU results; model objects remain local to this function.
    return {
        "branch": branch_name,
        "initial_validation_loss": baseline,
        "final_validation_loss": validation_rows[-1]["validation_loss"],
        "checkpoint": str(final_checkpoint),
    }
```

```python
branch_results = []

for branch_name in ("control", "stage_balanced"):
    result = train_branch(branch_name)
    branch_results.append(result)

    pd.DataFrame(branch_results).to_csv(
        PAIR_ROOT / "training_comparison.csv", index=False
    )

    # Release objects from the completed branch before loading the next.
    gc.collect()
    torch.cuda.empty_cache()

comparison = pd.DataFrame(branch_results)
display(comparison)

print("Both branches finished at 55,000 total steps.")
print("Results:", PAIR_ROOT)
```

```python
index_file = EVAL_ROOT / "completed_suite_paths.json"
completed_paths = (
    json.loads(index_file.read_text())
    if index_file.exists()
    else {}
)

summary = {}
task_rows = []

for suite in SUITES:
    saved = completed_paths.get(suite)
    result_file = Path(saved) if saved else None

    if result_file is None or not result_file.exists():
        candidates = [
            EVAL_ROOT / suite / "eval_info.json",
            *sorted(EVAL_ROOT.glob(
                f"{suite}_retry_*/eval_info.json"
            )),
        ]
        completed = [p for p in candidates if p.exists()]
        result_file = completed[-1] if completed else None

    if result_file is None:
        output_dir = EVAL_ROOT / suite

        if output_dir.exists():
            output_dir = EVAL_ROOT / (
                f"{suite}_retry_{datetime.now():%Y%m%d_%H%M%S_%f}"
            )

        log_path = EVAL_ROOT / f"{output_dir.name}_evaluation.log"

        command = [
            sys.executable, "-m", "lerobot.scripts.lerobot_eval",
            f"--policy.path={CHECKPOINT}",
            "--policy.device=cuda",
            "--policy.n_action_steps=1",
            "--env.type=libero",
            f"--env.task={suite}",
            "--env.max_parallel_tasks=1",
            "--eval.batch_size=1",
            "--eval.n_episodes=10",
            "--seed=1000",
            f"--rename_map={json.dumps(RENAME_MAP)}",
            f"--output_dir={output_dir}",
        ]

        print(f"\nStarting {suite}: 100 attempts", flush=True)
        print("Progress log:", log_path, flush=True)

        with log_path.open("w") as log:
            process = subprocess.Popen(
                command,
                cwd=PROJECT,
                env=env,
                stdin=subprocess.PIPE,
                stdout=log,
                stderr=subprocess.STDOUT,
                text=True,
            )

            try:
                process.communicate(input="N\n")
            except BaseException:
                process.terminate()
                try:
                    process.wait(timeout=15)
                except subprocess.TimeoutExpired:
                    process.kill()
                    process.wait()
                raise

        if process.returncode != 0:
            lines = log_path.read_text(
                errors="replace"
            ).splitlines()

            first_traceback = next(
                (i for i, line in enumerate(lines)
                 if line.startswith("Traceback")),
                max(0, len(lines) - 80),
            )
            print("\n".join(
                lines[first_traceback:first_traceback + 80]
            ))
            raise RuntimeError(f"Evaluation failed: {log_path}")

        result_file = output_dir / "eval_info.json"

    else:
        print(f"Reusing completed suite: {suite}", flush=True)

    result = json.loads(result_file.read_text())
    metrics = result["overall"]
    assert metrics["n_episodes"] == 100, metrics

    summary[suite] = metrics
    completed_paths[suite] = str(result_file)

    index_file.write_text(json.dumps(completed_paths, indent=2))
    (EVAL_ROOT / "summary.json").write_text(
        json.dumps(summary, indent=2)
    )

    for task in result["per_task"]:
        successes = task["metrics"]["successes"]

        task_rows.append({
            "suite": task["task_group"],
            "task_id": int(task["task_id"]),
            "episodes": len(successes),
            "successes": sum(bool(x) for x in successes),
            "success_percent": (
                100 * sum(bool(x) for x in successes)
                / len(successes)
            ),
        })

    pd.DataFrame(task_rows).to_csv(
        EVAL_ROOT / "per_task_results.csv", index=False
    )

    print(
        f"{suite}: original 50k={baseline_50k[suite]:.1f}% "
        f"→ stage-balanced 55k={metrics['pc_success']:.1f}%",
        flush=True,
    )

rows = [{
    "suite": suite,
    "episodes": summary[suite]["n_episodes"],
    "original_50k_percent": baseline_50k[suite],
    "stage_balanced_55k_percent": summary[suite]["pc_success"],
    "change_points": (
        summary[suite]["pc_success"] - baseline_50k[suite]
    ),
} for suite in SUITES]

table = pd.DataFrame(rows)
total = int(table["episodes"].sum())

overall = float(
    (table["stage_balanced_55k_percent"] * table["episodes"]).sum()
    / total
)

comparison = pd.concat([
    table,
    pd.DataFrame([{
        "suite": "OVERALL",
        "episodes": total,
        "original_50k_percent": 56.5,
        "stage_balanced_55k_percent": overall,
        "change_points": overall - 56.5,
    }]),
], ignore_index=True)

comparison.to_csv(EVAL_ROOT / "comparison.csv", index=False)

print("\nAll-suite comparison:")
display(comparison.round(2))

per_task = pd.DataFrame(task_rows)
print("\nLIBERO_10 task results:")
display(
    per_task.loc[per_task["suite"] == "libero_10"]
    .sort_values("task_id")
    .round(2)
)

print("Total attempts:", total)
print("Results and videos:", EVAL_ROOT)
```

# Gripper-Weighted Loss

```python
import gc
import json
import math
import random
import time
import warnings
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from IPython.display import display
from tqdm.auto import tqdm

from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.configs.policies import PreTrainedConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy
from lerobot.policies.factory import make_pre_post_processors

PROJECT = Path("/home/hasith/Desktop/SmolVLA")

SOURCE = (
    PROJECT
    / "outputs/finetuning/base16_corrected_from_start_100k_20261122_154723"
)
INTERFACE = SOURCE / "checkpoints/000000/pretrained_model"

BASE_REVISION = "5e8d12a6e2975b0e5e5fce7c8caf47c371d257b6"
BASE_PATH = (
    Path.home()
    / ".cache/huggingface/hub/models--lerobot--smolvla_base/snapshots"
    / BASE_REVISION
)

STEPS = 10_000
BATCH_SIZE = 16
SAVE_EVERY = 5_000
GRIPPER_WEIGHT = 2.0

SEED = 1000
PEAK_LR = 1e-4
END_LR = 2.5e-6
WARMUP = 1_000

assert torch.cuda.is_available()
assert (BASE_PATH / "model.safetensors").exists()
assert (INTERFACE / "config.json").exists()

split = json.loads((SOURCE / "episode_split.json").read_text())
val_indices = json.loads(
    (SOURCE / "validation_indices.json").read_text()
)
experiment = json.loads((SOURCE / "experiment.json").read_text())

assert split["train"] and split["validation"]
assert not set(split["train"]) & set(split["validation"])

RUN = PROJECT / "outputs/finetuning" / (
    f"base16_gripper_x2_10k_bs16_{datetime.now():%Y%m%d_%H%M%S}"
)
RUN.mkdir(parents=True, exist_ok=False)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

warnings.filterwarnings(
    "ignore",
    message=r"The video decoding and encoding capabilities of torchvision.*",
    category=UserWarning,
)

# LIBERO configuration only; weights are loaded from BASE_PATH below.
cfg = PreTrainedConfig.from_pretrained(str(INTERFACE))
assert isinstance(cfg, SmolVLAConfig)
assert cfg.num_vlm_layers == 16

cfg.device = "cuda"
cfg.push_to_hub = False
cfg.freeze_vision_encoder = True
cfg.train_expert_only = True
cfg.train_state_proj = True
cfg.chunk_size = 50
cfg.n_action_steps = 1

policy = SmolVLAPolicy.from_pretrained(
    str(BASE_PATH),
    config=cfg,
    strict=True,
).to("cuda")

# Reuse LIBERO normalization calculated from training episodes only.
preprocess, postprocess = make_pre_post_processors(
    policy.config,
    pretrained_path=str(INTERFACE),
    preprocessor_overrides={
        "device_processor": {"device": "cuda"},
        "rename_observations_processor": {
            "rename_map": {
                "observation.images.image": "observation.images.camera1",
                "observation.images.image2": "observation.images.camera2",
            }
        },
    },
)

assert tuple(policy.config.action_feature.shape) == (7,)
assert not policy.config.adapt_to_pi_aloha
assert all(
    not p.requires_grad
    for p in policy.model.vlm_with_expert.vlm.parameters()
)

trainable = [
    p for p in policy.parameters()
    if p.requires_grad
]

for filename in (
    "episode_split.json",
    "validation_indices.json",
    "training_statistics.json",
):
    (RUN / filename).write_text(
        (SOURCE / filename).read_text()
    )

settings = {
    "model_id": "lerobot/smolvla_base",
    "model_revision": BASE_REVISION,
    "starting_weights": str(BASE_PATH),
    "dataset_id": experiment["dataset_id"],
    "dataset_revision": experiment["dataset_revision"],
    "steps": STEPS,
    "batch_size": BATCH_SIZE,
    "save_every": SAVE_EVERY,
    "seed": SEED,
    "peak_lr": PEAK_LR,
    "end_lr": END_LR,
    "warmup": WARMUP,
    "gripper_weight": GRIPPER_WEIGHT,
    "objective": (
        "weighted flow matching; valid timesteps; "
        "7 real action dimensions"
    ),
    "optimizer": "fresh AdamW",
    "preprocessing_source": str(INTERFACE),
}

(RUN / "experiment.json").write_text(
    json.dumps(settings, indent=2)
)

print("Loaded pretrained BASE weights:", BASE_PATH)
print(
    "Training / validation episodes:",
    len(split["train"]), "/", len(split["validation"]),
)
print("Steps:", STEPS)
print("Batch size:", BATCH_SIZE)
print("Output:", RUN)
```

```python
from torch.utils.data import DataLoader, Subset

from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

meta = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)

common = {
    "repo_id": experiment["dataset_id"],
    "revision": experiment["dataset_revision"],
    "delta_timestamps": resolve_delta_timestamps(
        policy.config, meta
    ),
    "video_backend": "pyav",
}

train_dataset = LeRobotDataset(
    **common,
    episodes=split["train"],
)
val_dataset = LeRobotDataset(
    **common,
    episodes=split["validation"],
)

assert set(train_dataset.episodes) == set(split["train"])
assert set(val_dataset.episodes) == set(split["validation"])
assert min(val_indices) >= 0
assert max(val_indices) < len(val_dataset)

loader_generator = torch.Generator().manual_seed(SEED)
val_generator = torch.Generator().manual_seed(SEED + 1)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    drop_last=True,
    num_workers=0,
    generator=loader_generator,
)

val_loader = DataLoader(
    Subset(val_dataset, val_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    generator=val_generator,
)


def loss_components(batch):
    images, image_masks = policy.prepare_images(batch)

    element_losses = policy.model.forward(
        images,
        image_masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        policy.prepare_state(batch),
        policy.prepare_action(batch),
    )

    # Six motion coordinates + one gripper coordinate.
    real = element_losses[..., :7]


    #Padded timesteps do not contribute to the loss.
    valid = ~batch["action_is_pad"].to(
        device=real.device,
        dtype=torch.bool,
    )
    assert valid.shape == real.shape[:2]

    count = valid.sum()
    if count.item() == 0:
        raise RuntimeError("No valid action targets in batch.")

    coordinate_sums = (
        real * valid.unsqueeze(-1).to(real.dtype)
    ).sum(dim=(0, 1))

    weights = real.new_tensor(
        [1, 1, 1, 1, 1, 1, GRIPPER_WEIGHT]
    )

    weighted = (
        (coordinate_sums * weights).sum()
        / (count * weights.sum())
    )

    return weighted, coordinate_sums, count


@torch.no_grad()
def validate():
    was_training = policy.training
    policy.eval()

    sums = torch.zeros(
        7, device="cuda", dtype=torch.float64
    )
    total_valid = 0

    try:
        with torch.random.fork_rng(
            devices=list(range(torch.cuda.device_count()))
        ):
            torch.manual_seed(SEED)
            torch.cuda.manual_seed_all(SEED)

            for raw in val_loader:
                batch = preprocess(dict(raw))
                _, coordinate_sums, count = loss_components(batch)

                sums += coordinate_sums.double()
                total_valid += int(count.item())
    finally:
        policy.train(was_training)

    assert total_valid > 0

    means = sums / total_valid
    weights = means.new_tensor(
        [1, 1, 1, 1, 1, 1, GRIPPER_WEIGHT]
    )

    return {
        "ordinary_validation_loss": means.mean().item(),
        "weighted_validation_loss": (
            (means * weights).sum() / weights.sum()
        ).item(),
        "motion_loss": means[:6].mean().item(),
        "gripper_loss": means[6].item(),
    }


print("Training frames:", len(train_dataset))
print("Fixed validation samples:", len(val_indices))
print(
    "Approximate training epochs:",
    round(STEPS / len(train_loader), 3),
)
```

```python
optimizer = torch.optim.AdamW(
    trainable,
    lr=PEAK_LR,
    betas=tuple(policy.config.optimizer_betas),
    eps=policy.config.optimizer_eps,
    weight_decay=policy.config.optimizer_weight_decay,
)


def learning_rate(step):
    if step <= WARMUP:
        return PEAK_LR * step / WARMUP

    fraction = (step - WARMUP) / (STEPS - WARMUP)

    return END_LR + 0.5 * (PEAK_LR - END_LR) * (
        1 + math.cos(math.pi * fraction)
    )


def save_checkpoint(step):
    destination = RUN / "checkpoints" / f"{step:06d}"
    model_dir = destination / "pretrained_model"
    model_dir.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(model_dir)
    preprocess.save_pretrained(model_dir)
    postprocess.save_pretrained(model_dir)

    torch.save(
        {
            "step": step,
            "optimizer": optimizer.state_dict(),
            "torch_rng": torch.get_rng_state(),
            "cuda_rng": torch.cuda.get_rng_state_all(),
            "python_rng": random.getstate(),
            "numpy_rng": np.random.get_state(),
            "loader_generator_rng": (
                loader_generator.get_state()
            ),
            "schedule": {
                "steps": STEPS,
                "warmup": WARMUP,
                "peak_lr": PEAK_LR,
                "end_lr": END_LR,
            },
            "gripper_weight": GRIPPER_WEIGHT,
            "objective": settings["objective"],
        },
        destination / "training_state.pt",
    )

    return model_dir


initial_checkpoint = save_checkpoint(0)

validation_rows = [{
    "step": 0,
    **validate(),
    "checkpoint": str(initial_checkpoint),
}]
training_rows = []

pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv",
    index=False,
)

print(
    "Starting validation:",
    validation_rows[0],
    flush=True,
)

iterator = iter(train_loader)
policy.train()
started = time.monotonic()

for step in tqdm(
    range(1, STEPS + 1),
    desc="Base → 10k: gripper ×2",
):
    lr = learning_rate(step)

    for group in optimizer.param_groups:
        group["lr"] = lr

    try:
        raw = next(iterator)
    except StopIteration:
        iterator = iter(train_loader)
        raw = next(iterator)

    batch = preprocess(dict(raw))
    optimizer.zero_grad(set_to_none=True)

    loss, coordinate_sums, valid_count = loss_components(batch)

    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite loss at step {step}")

    loss.backward()

    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable,
        policy.config.optimizer_grad_clip_norm,
        error_if_nonfinite=True,
    )
    optimizer.step()

    if step == 1 or step % 100 == 0:
        training_rows.append({
            "step": step,
            "weighted_training_loss": loss.item(),
            "grad_norm": float(grad_norm),
            "lr": lr,
            "elapsed_seconds": (
                time.monotonic() - started
            ),
        })

        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv",
            index=False,
        )

    del loss, coordinate_sums, valid_count, batch, raw

    if step % SAVE_EVERY == 0:
        model_dir = save_checkpoint(step)
        scores = validate()

        validation_rows.append({
            "step": step,
            **scores,
            "checkpoint": str(model_dir),
        })

        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv",
            index=False,
        )

        tqdm.write(
            f"Step {step}: "
            f"ordinary validation="
            f"{scores['ordinary_validation_loss']:.6f}, "
            f"weighted validation="
            f"{scores['weighted_validation_loss']:.6f}, "
            f"gripper={scores['gripper_loss']:.6f}"
        )

FINAL_CHECKPOINT = (
    RUN / "checkpoints/010000/pretrained_model"
)

optimizer.zero_grad(set_to_none=True)
del iterator
gc.collect()
torch.cuda.empty_cache()

print("Finished: 10,000 steps from the pretrained base.")
print("Final checkpoint:", FINAL_CHECKPOINT)
display(pd.DataFrame(validation_rows))
```

```python
import gc
import json
import math
import random
import time
import warnings
from datetime import datetime
from pathlib import Path

import numpy as np
import pandas as pd
import torch
from IPython.display import display
from tqdm.auto import tqdm
from torchvision.transforms.v2 import functional as TF

from lerobot.policies.smolvla.configuration_smolvla import SmolVLAConfig
from lerobot.configs.policies import PreTrainedConfig
from lerobot.policies.smolvla.modeling_smolvla import SmolVLAPolicy
from lerobot.policies.factory import make_pre_post_processors

PROJECT = Path("/home/hasith/Desktop/SmolVLA")
SOURCE = (
    PROJECT / "outputs/finetuning"
    / "base16_corrected_from_start_100k_20261122_154723"
)
INTERFACE = SOURCE / "checkpoints/000000/pretrained_model"

BASE_REVISION = "5e8d12a6e2975b0e5e5fce7c8caf47c371d257b6"
BASE_PATH = (
    Path.home()
    / ".cache/huggingface/hub/models--lerobot--smolvla_base/snapshots"
    / BASE_REVISION
)

STEPS = 10_000
BATCH_SIZE = 16
SAVE_EVERY = 5_000

SEED = 1000
PEAK_LR = 1e-4
END_LR = 2.5e-6
WARMUP = 1_000

# Change to False for a matched run without augmentation.
AUGMENT = True
BRIGHTNESS_RANGE = (0.9, 1.1)
CONTRAST_RANGE = (0.9, 1.1)

assert torch.cuda.is_available()
assert (BASE_PATH / "model.safetensors").exists()
assert (INTERFACE / "config.json").exists()

split = json.loads((SOURCE / "episode_split.json").read_text())
val_indices = json.loads(
    (SOURCE / "validation_indices.json").read_text()
)
experiment = json.loads((SOURCE / "experiment.json").read_text())

assert split["train"] and split["validation"]
assert not set(split["train"]) & set(split["validation"])

label = "image_aug" if AUGMENT else "no_aug_control"
RUN = PROJECT / "outputs/finetuning" / (
    f"base16_{label}_10k_bs16_{datetime.now():%Y%m%d_%H%M%S}"
)
RUN.mkdir(parents=True, exist_ok=False)

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

warnings.filterwarnings(
    "ignore",
    message=r"The video decoding and encoding capabilities of torchvision.*",
    category=UserWarning,
)

# Reuse only the LIBERO interface configuration.
cfg = PreTrainedConfig.from_pretrained(str(INTERFACE))
assert isinstance(cfg, SmolVLAConfig)
assert cfg.num_vlm_layers == 16

cfg.device = "cuda"
cfg.push_to_hub = False
cfg.freeze_vision_encoder = True
cfg.train_expert_only = True
cfg.train_state_proj = True
cfg.chunk_size = 50
cfg.n_action_steps = 1

# Load pretrained BASE weights, not the old fine-tuned weights.
policy = SmolVLAPolicy.from_pretrained(
    str(BASE_PATH),
    config=cfg,
    strict=True,
).to("cuda")

# Reuse LIBERO normalization calculated from training episodes only.
preprocess, postprocess = make_pre_post_processors(
    policy.config,
    pretrained_path=str(INTERFACE),
    preprocessor_overrides={
        "device_processor": {"device": "cuda"},
        "rename_observations_processor": {
            "rename_map": {
                "observation.images.image":
                    "observation.images.camera1",
                "observation.images.image2":
                    "observation.images.camera2",
            }
        },
    },
)

assert tuple(policy.config.action_feature.shape) == (7,)
assert not policy.config.adapt_to_pi_aloha
assert all(
    not p.requires_grad
    for p in policy.model.vlm_with_expert.vlm.parameters()
)

trainable = [
    p for p in policy.parameters() if p.requires_grad
]

for filename in (
    "episode_split.json",
    "validation_indices.json",
    "training_statistics.json",
):
    (RUN / filename).write_text(
        (SOURCE / filename).read_text()
    )

settings = {
    "starting_weights": str(BASE_PATH),
    "model_revision": BASE_REVISION,
    "dataset_id": experiment["dataset_id"],
    "dataset_revision": experiment["dataset_revision"],
    "steps": STEPS,
    "batch_size": BATCH_SIZE,
    "seed": SEED,
    "peak_lr": PEAK_LR,
    "end_lr": END_LR,
    "warmup": WARMUP,
    "optimizer": "fresh AdamW",
    "objective": "ordinary flow matching; valid steps; 7 real dimensions",
    "augmentation_enabled": AUGMENT,
    "brightness_range": BRIGHTNESS_RANGE,
    "contrast_range": CONTRAST_RANGE,
    "augmentation_training_only": True,
}
(RUN / "experiment.json").write_text(
    json.dumps(settings, indent=2)
)

print("Starting from pretrained base:", BASE_PATH)
print("Training / validation episodes:",
      len(split["train"]), "/", len(split["validation"]))
print("Augmentation enabled:", AUGMENT)
print("Output:", RUN)
```

```python
from torch.utils.data import DataLoader, Subset
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

meta = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)

common = {
    "repo_id": experiment["dataset_id"],
    "revision": experiment["dataset_revision"],
    "delta_timestamps": resolve_delta_timestamps(
        policy.config, meta
    ),
    "video_backend": "pyav",
}

train_dataset = LeRobotDataset(
    **common, episodes=split["train"]
)
val_dataset = LeRobotDataset(
    **common, episodes=split["validation"]
)

assert set(train_dataset.episodes) == set(split["train"])
assert set(val_dataset.episodes) == set(split["validation"])
assert min(val_indices) >= 0
assert max(val_indices) < len(val_dataset)

loader_generator = torch.Generator().manual_seed(SEED)
val_generator = torch.Generator().manual_seed(SEED + 1)

# Augmentation has its own RNG, separate from training noise.
augmentation_generator = torch.Generator().manual_seed(SEED + 2)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    drop_last=True,
    num_workers=0,
    generator=loader_generator,
)

val_loader = DataLoader(
    Subset(val_dataset, val_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    generator=val_generator,
)


def augment_images(raw):
    result = dict(raw)
    if not AUGMENT:
        return result

    image_keys = [
        key for key in result
        if key.startswith("observation.images.")
    ]
    assert image_keys, "No camera images found."

    batch_size = result[image_keys[0]].shape[0]
    random_factors = torch.rand(
        batch_size, 2,
        generator=augmentation_generator,
    )

    brightness = (
        BRIGHTNESS_RANGE[0]
        + random_factors[:, 0]
        * (BRIGHTNESS_RANGE[1] - BRIGHTNESS_RANGE[0])
    )
    contrast = (
        CONTRAST_RANGE[0]
        + random_factors[:, 1]
        * (CONTRAST_RANGE[1] - CONTRAST_RANGE[0])
    )

    for key in image_keys:
        images = result[key]
        assert images.is_floating_point()
        assert torch.isfinite(images).all()
        assert images.min() >= 0 and images.max() <= 1

        augmented = []
        for i in range(batch_size):
            image = TF.adjust_brightness(
                images[i], float(brightness[i])
            )
            image = TF.adjust_contrast(
                image, float(contrast[i])
            )
            augmented.append(image)

        result[key] = torch.stack(augmented)

    return result


def corrected_loss(batch):
    images, image_masks = policy.prepare_images(batch)

    element_losses = policy.model.forward(
        images,
        image_masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        policy.prepare_state(batch),
        policy.prepare_action(batch),
    )

    real = element_losses[..., :7]
    valid = ~batch["action_is_pad"].to(
        device=real.device, dtype=torch.bool
    )
    assert valid.shape == real.shape[:2]

    count = valid.sum()
    if count.item() == 0:
        raise RuntimeError("No valid targets in batch.")

    # Equal weights for all seven real action coordinates.
    return (
        real * valid.unsqueeze(-1).to(real.dtype)
    ).sum() / (count * 7)


@torch.no_grad()
def validate():
    was_training = policy.training
    policy.eval()
    total_loss = 0.0
    total_targets = 0

    try:
        with torch.random.fork_rng(
            devices=list(range(torch.cuda.device_count()))
        ):
            torch.manual_seed(SEED)
            torch.cuda.manual_seed_all(SEED)

            for raw in val_loader:
                # Validation images are never augmented.
                batch = preprocess(dict(raw))
                loss = corrected_loss(batch)

                if not torch.isfinite(loss):
                    raise RuntimeError("Non-finite validation loss.")

                targets = int(
                    (~batch["action_is_pad"].bool()).sum().item()
                ) * 7

                total_loss += loss.item() * targets
                total_targets += targets
    finally:
        policy.train(was_training)

    assert total_targets > 0
    return total_loss / total_targets


print("Training frames:", len(train_dataset))
print("Validation samples:", len(val_indices))
print("Approximate epochs:",
      round(STEPS / len(train_loader), 3))
```

```python
optimizer = torch.optim.AdamW(
    trainable,
    lr=PEAK_LR,
    betas=tuple(policy.config.optimizer_betas),
    eps=policy.config.optimizer_eps,
    weight_decay=policy.config.optimizer_weight_decay,
)


def learning_rate(step):
    if step <= WARMUP:
        return PEAK_LR * step / WARMUP

    fraction = (step - WARMUP) / (STEPS - WARMUP)
    return END_LR + 0.5 * (PEAK_LR - END_LR) * (
        1 + math.cos(math.pi * fraction)
    )


def save_checkpoint(step):
    destination = RUN / "checkpoints" / f"{step:06d}"
    model_dir = destination / "pretrained_model"
    model_dir.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(model_dir)
    preprocess.save_pretrained(model_dir)
    postprocess.save_pretrained(model_dir)

    torch.save({
        "step": step,
        "optimizer": optimizer.state_dict(),
        "torch_rng": torch.get_rng_state(),
        "cuda_rng": torch.cuda.get_rng_state_all(),
        "python_rng": random.getstate(),
        "numpy_rng": np.random.get_state(),
        "loader_generator_rng": loader_generator.get_state(),
        "augmentation_generator_rng":
            augmentation_generator.get_state(),
        "schedule": {
            "steps": STEPS,
            "warmup": WARMUP,
            "peak_lr": PEAK_LR,
            "end_lr": END_LR,
        },
        "settings": settings,
    }, destination / "training_state.pt")

    return model_dir


initial_checkpoint = save_checkpoint(0)

validation_rows = [{
    "step": 0,
    "validation_loss": validate(),
    "checkpoint": str(initial_checkpoint),
}]
training_rows = []

pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv", index=False
)
print("Starting validation:",
      validation_rows[0]["validation_loss"], flush=True)

iterator = iter(train_loader)
policy.train()
started = time.monotonic()

for step in tqdm(
    range(1, STEPS + 1),
    desc="Base → 10k: image augmentation",
):
    lr = learning_rate(step)
    for group in optimizer.param_groups:
        group["lr"] = lr

    try:
        raw = next(iterator)
    except StopIteration:
        iterator = iter(train_loader)
        raw = next(iterator)

    augmented = augment_images(raw)
    batch = preprocess(augmented)

    optimizer.zero_grad(set_to_none=True)
    loss = corrected_loss(batch)

    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite loss at step {step}")

    loss.backward()
    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable,
        policy.config.optimizer_grad_clip_norm,
        error_if_nonfinite=True,
    )
    optimizer.step()

    if step == 1 or step % 100 == 0:
        training_rows.append({
            "step": step,
            "loss": loss.item(),
            "grad_norm": float(grad_norm),
            "lr": lr,
            "elapsed_seconds": time.monotonic() - started,
        })
        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv", index=False
        )

    del loss, batch, augmented, raw

    if step % SAVE_EVERY == 0:
        model_dir = save_checkpoint(step)
        score = validate()

        validation_rows.append({
            "step": step,
            "validation_loss": score,
            "checkpoint": str(model_dir),
        })  
        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv", index=False
        )

        tqdm.write(
            f"Step {step}: validation={score:.6f}; "
            f"saved {model_dir}"
        )

FINAL_CHECKPOINT = RUN / "checkpoints/010000/pretrained_model"

optimizer.zero_grad(set_to_none=True)
del iterator
gc.collect()
torch.cuda.empty_cache()

print("Finished: 10,000 steps from pretrained base.")
print("Final checkpoint:", FINAL_CHECKPOINT)
display(pd.DataFrame(validation_rows))
```

```python
from torch.utils.data import DataLoader, Subset, WeightedRandomSampler
from lerobot.datasets.lerobot_dataset import (
    LeRobotDataset,
    LeRobotDatasetMetadata,
)
from lerobot.datasets.factory import resolve_delta_timestamps

meta = LeRobotDatasetMetadata(
    experiment["dataset_id"],
    revision=experiment["dataset_revision"],
)
common = {
    "repo_id": experiment["dataset_id"],
    "revision": experiment["dataset_revision"],
    "delta_timestamps": resolve_delta_timestamps(policy.config, meta),
    "video_backend": "pyav",
}

train_dataset = LeRobotDataset(**common, episodes=split["train"])
val_dataset = LeRobotDataset(**common, episodes=split["validation"])

assert set(train_dataset.episodes) == set(split["train"])
assert set(val_dataset.episodes) == set(split["validation"])
assert min(val_indices) >= 0
assert max(val_indices) < len(val_dataset)

# Read identifiers and actions without decoding images.
table = (
    train_dataset.hf_dataset
    .select_columns([
        "episode_index", "frame_index", "task_index", "action"
    ])
    .to_pandas()
    .reset_index(drop=True)
)
assert len(table) == len(train_dataset)

actions = np.stack(table["action"].to_numpy())
assert actions.shape[1] == 7
gripper = actions[:, -1]
assert np.all(np.isclose(np.abs(gripper), 1.0)), (
    "Unexpected gripper encoding; check before detecting events."
)

event_mask = np.zeros(len(table), dtype=bool)

# Keep event windows within their own episode.
for _, group in table.groupby("episode_index", sort=False):
    ordered = group.sort_values("frame_index")
    indices = ordered.index.to_numpy()
    frames = ordered["frame_index"].to_numpy()
    commands = gripper[indices]

    changes = np.flatnonzero(
        np.abs(np.diff(commands)) > 1.0
    ) + 1

    for position in changes:
        nearby = np.abs(frames - frames[position]) <= EVENT_RADIUS
        event_mask[indices[nearby]] = True

assert event_mask.any(), "No gripper-change windows detected."

task_ids = table["task_index"].to_numpy(dtype=np.int64)
n = len(table)
weights = np.zeros(n, dtype=np.float64)
sampling_rows = []

mix = EVENT_MIX if STAGE_BALANCED else 0.0

for task_id in sorted(np.unique(task_ids)):
    task_mask = task_ids == task_id
    event_indices = task_mask & event_mask

    count = int(task_mask.sum())
    event_count = int(event_indices.sum())
    task_probability = count / n

    if event_count:
        weights[task_mask] = task_probability * (1 - mix) / count
        weights[event_indices] += task_probability * mix / event_count
    else:
        weights[task_mask] = task_probability / count

    assert np.isclose(weights[task_mask].sum(), task_probability)

    sampling_rows.append({
        "dataset_task_id": int(task_id),
        "training_frames": count,
        "event_window_frames": event_count,
        "task_probability": task_probability,
        "uniform_event_fraction": event_count / count,
        "sampled_event_fraction":
            weights[event_indices].sum() / task_probability,
    })

assert np.isclose(weights.sum(), 1.0)

sampling_report = pd.DataFrame(sampling_rows)
sampling_report.to_csv(RUN / "sampling_report.csv", index=False)

sampler_generator = torch.Generator().manual_seed(SEED)
loader_generator = torch.Generator().manual_seed(SEED + 1)
val_generator = torch.Generator().manual_seed(SEED + 2)

sampler = WeightedRandomSampler(
    torch.from_numpy(weights),
    num_samples=STEPS * BATCH_SIZE,
    replacement=True,
    generator=sampler_generator,
)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    drop_last=True,
    num_workers=0,
    generator=loader_generator,
)
val_loader = DataLoader(
    Subset(val_dataset, val_indices),
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    generator=val_generator,
)

del table, actions, gripper

print("Training frames:", n)
print("Training / validation episodes:",
      len(split["train"]), "/", len(split["validation"]))
print("Sample presentations:", STEPS * BATCH_SIZE)
display(sampling_report)
```

```python
def corrected_loss(batch):
    images, masks = policy.prepare_images(batch)
    losses = policy.model.forward(
        images,
        masks,
        batch["observation.language.tokens"],
        batch["observation.language.attention_mask"],
        policy.prepare_state(batch),
        policy.prepare_action(batch),
    )

    real = losses[..., :7]
    valid = ~batch["action_is_pad"].to(
        device=real.device, dtype=torch.bool
    )
    assert valid.shape == real.shape[:2]

    denominator = valid.sum() * 7
    if denominator.item() == 0:
        raise RuntimeError("No valid action targets.")

    return (
        real * valid.unsqueeze(-1).to(real.dtype)
    ).sum() / denominator


@torch.no_grad()
def validate():
    was_training = policy.training
    policy.eval()
    numerator, denominator = 0.0, 0

    try:
        with torch.random.fork_rng(
            devices=list(range(torch.cuda.device_count()))
        ):
            torch.manual_seed(SEED)
            torch.cuda.manual_seed_all(SEED)

            for raw in val_loader:
                batch = preprocess(dict(raw))
                value = corrected_loss(batch)
                if not torch.isfinite(value):
                    raise RuntimeError("Non-finite validation loss.")

                count = int(
                    (~batch["action_is_pad"].bool()).sum().item()
                ) * 7
                numerator += value.item() * count
                denominator += count
    finally:
        policy.train(was_training)

    assert denominator > 0
    return numerator / denominator


optimizer = torch.optim.AdamW(
    trainable,
    lr=PEAK_LR,
    betas=tuple(policy.config.optimizer_betas),
    eps=policy.config.optimizer_eps,
    weight_decay=policy.config.optimizer_weight_decay,
)


def learning_rate(step):
    if step <= WARMUP:
        return PEAK_LR * step / WARMUP
    fraction = (step - WARMUP) / (STEPS - WARMUP)
    return END_LR + 0.5 * (PEAK_LR - END_LR) * (
        1 + math.cos(math.pi * fraction)
    )


def save_checkpoint(step):
    folder = RUN / "checkpoints" / f"{step:06d}"
    destination = folder / "pretrained_model"
    destination.mkdir(parents=True, exist_ok=False)

    policy.save_pretrained(destination)
    preprocess.save_pretrained(destination)
    postprocess.save_pretrained(destination)

    torch.save({
        "step": step,
        "optimizer": optimizer.state_dict(),
        "torch_rng": torch.get_rng_state(),
        "cuda_rng": torch.cuda.get_rng_state_all(),
        "python_rng": random.getstate(),
        "numpy_rng": np.random.get_state(),
        "sampler_generator_rng": sampler_generator.get_state(),
        "loader_generator_rng": loader_generator.get_state(),
        "settings": settings,
    }, folder / "training_state.pt")

    return destination


initial = save_checkpoint(0)
validation_rows = [{
    "step": 0,
    "validation_loss": validate(),
    "checkpoint": str(initial),
}]
training_rows = []

pd.DataFrame(validation_rows).to_csv(
    RUN / "validation.csv", index=False
)
print("Starting validation:",
      validation_rows[0]["validation_loss"], flush=True)

policy.train()
started = time.monotonic()

for step, raw in enumerate(
    tqdm(train_loader, desc=f"{label}: base → 10k"),
    start=1,
):
    lr = learning_rate(step)
    for group in optimizer.param_groups:
        group["lr"] = lr

    batch = preprocess(dict(raw))
    optimizer.zero_grad(set_to_none=True)
    loss = corrected_loss(batch)

    if not torch.isfinite(loss):
        raise RuntimeError(f"Non-finite loss at step {step}")

    loss.backward()
    grad_norm = torch.nn.utils.clip_grad_norm_(
        trainable,
        policy.config.optimizer_grad_clip_norm,
        error_if_nonfinite=True,
    )
    optimizer.step()

    if step == 1 or step % 100 == 0:
        training_rows.append({
            "step": step,
            "loss": loss.item(),
            "grad_norm": float(grad_norm),
            "lr": lr,
            "elapsed_seconds": time.monotonic() - started,
        })
        pd.DataFrame(training_rows).to_csv(
            RUN / "training.csv", index=False
        )

    del loss, batch, raw

    if step % SAVE_EVERY == 0:
        model_dir = save_checkpoint(step)
        score = validate()

        validation_rows.append({
            "step": step,
            "validation_loss": score,
            "checkpoint": str(model_dir),
        })
        pd.DataFrame(validation_rows).to_csv(
            RUN / "validation.csv", index=False
        )
        tqdm.write(
            f"Step {step}: validation={score:.6f}; saved {model_dir}"
        )

FINAL_CHECKPOINT = RUN / "checkpoints/010000/pretrained_model"
optimizer.zero_grad(set_to_none=True)

print("Finished: 10,000 steps from pretrained base.")
print("Final checkpoint:", FINAL_CHECKPOINT)
display(pd.DataFrame(validation_rows))
```

```python
import json
import os
import subprocess
import sys
from datetime import datetime
from pathlib import Path

import pandas as pd
from IPython.display import display

PROJECT = Path("/home/hasith/Desktop/SmolVLA")

TRAIN_RUN = (
    PROJECT / "outputs/finetuning"
    / "base16_stage_balanced_10k_bs16_20261125_025100"
)
CHECKPOINT = TRAIN_RUN / "checkpoints/010000/pretrained_model"

assert (CHECKPOINT / "model.safetensors").exists(), (
    f"10k checkpoint not saved yet: {CHECKPOINT}"
)

SUITES = [
    "libero_spatial",
    "libero_object",
    "libero_goal",
    "libero_10",
]

RENAME_MAP = {
    "observation.images.image": "observation.images.camera1",
    "observation.images.image2": "observation.images.camera2",
}

EVAL_ROOT = PROJECT / "outputs/eval" / (
    f"stage_balanced_base_10k_400_{datetime.now():%Y%m%d_%H%M%S}"
)
EVAL_ROOT.mkdir(parents=True, exist_ok=False)

env = {
    **os.environ,
    "MUJOCO_GL": "egl",
    "PYOPENGL_PLATFORM": "egl",
    "PYTHONUNBUFFERED": "1",
    "HF_HUB_DOWNLOAD_TIMEOUT": "120",
    "HF_HUB_ETAG_TIMEOUT": "120",
}

(EVAL_ROOT / "evaluation_settings.json").write_text(
    json.dumps({
        "checkpoint": str(CHECKPOINT),
        "seed": 1000,
        "episodes_per_task": 10,
        "n_action_steps": 1,
        "rename_map": RENAME_MAP,
    }, indent=2)
)

print("Testing:", CHECKPOINT)
print("Results:", EVAL_ROOT)
```

```python
baseline_10k = {
    "libero_spatial": 58.0,
    "libero_object": 70.0,
    "libero_goal": 49.0,
    "libero_10": 29.0,
}

gripper_10k = {
    "libero_spatial": 61.0,
    "libero_object": 48.0,
    "libero_goal": 60.0,
    "libero_10": 23.0,
}

index_file = EVAL_ROOT / "completed_suite_paths.json"
completed_paths = (
    json.loads(index_file.read_text())
    if index_file.exists()
    else {}
)

summary = {}

for suite in SUITES:
    saved = completed_paths.get(suite)
    result_file = Path(saved) if saved else None

    if result_file is None or not result_file.exists():
        candidates = [
            EVAL_ROOT / suite / "eval_info.json",
            *sorted(EVAL_ROOT.glob(
                f"{suite}_retry_*/eval_info.json"
            )),
        ]
        completed = [p for p in candidates if p.exists()]
        result_file = completed[-1] if completed else None

    if result_file is None:
        output_dir = EVAL_ROOT / suite

        if output_dir.exists():
            output_dir = EVAL_ROOT / (
                f"{suite}_retry_{datetime.now():%Y%m%d_%H%M%S_%f}"
            )

        log_path = EVAL_ROOT / f"{output_dir.name}_evaluation.log"

        command = [
            sys.executable, "-m", "lerobot.scripts.lerobot_eval",
            f"--policy.path={CHECKPOINT}",
            "--policy.device=cuda",
            "--policy.n_action_steps=1",
            "--env.type=libero",
            f"--env.task={suite}",
            "--env.max_parallel_tasks=1",
            "--eval.batch_size=1",
            "--eval.n_episodes=10",
            "--seed=1000",
            f"--rename_map={json.dumps(RENAME_MAP)}",
            f"--output_dir={output_dir}",
        ]

        print(f"\nStarting {suite}: 100 attempts", flush=True)
        print("Progress log:", log_path, flush=True)

        with log_path.open("w") as log:
            process = subprocess.Popen(
                command,
                cwd=str(PROJECT),
                env=env,
                stdin=subprocess.PIPE,
                stdout=log,
                stderr=subprocess.STDOUT,
                text=True,
            )

            try:
                process.communicate(input="N\n")
            except BaseException:
                process.terminate()
                try:
                    process.wait(timeout=15)
                except subprocess.TimeoutExpired:
                    process.kill()
                    process.wait()
                raise

        if process.returncode != 0:
            lines = log_path.read_text(
                errors="replace"
            ).splitlines()

            first_traceback = next(
                (
                    i for i, line in enumerate(lines)
                    if line.startswith("Traceback")
                ),
                max(0, len(lines) - 80),
            )

            print("\n".join(
                lines[first_traceback:first_traceback + 100]
            ))
            raise RuntimeError(
                f"Evaluation failed. Full log: {log_path}"
            )

        result_file = output_dir / "eval_info.json"
    else:
        print(f"Reusing completed suite: {suite}", flush=True)

    metrics = json.loads(result_file.read_text())["overall"]
    assert metrics["n_episodes"] == 100, metrics

    summary[suite] = metrics
    completed_paths[suite] = str(result_file)

    index_file.write_text(
        json.dumps(completed_paths, indent=2)
    )
    (EVAL_ROOT / "summary.json").write_text(
        json.dumps(summary, indent=2)
    )

    print(
        f"{suite}: ordinary 10k={baseline_10k[suite]:.1f}% "
        f"→ stage-balanced 10k={metrics['pc_success']:.1f}%",
        flush=True,
    )

table = pd.DataFrame([{
    "suite": suite,
    "episodes": summary[suite]["n_episodes"],
    "ordinary_10k_percent": baseline_10k[suite],
    "gripper_x2_10k_percent": gripper_10k[suite],
    "stage_balanced_10k_percent": summary[suite]["pc_success"],
    "change_vs_ordinary_points": (
        summary[suite]["pc_success"] - baseline_10k[suite]
    ),
} for suite in SUITES])

total = int(table["episodes"].sum())
overall = float(
    (
        table["stage_balanced_10k_percent"]
        * table["episodes"]
    ).sum() / total
)

comparison = pd.concat([
    table,
    pd.DataFrame([{
        "suite": "OVERALL",
        "episodes": total,
        "ordinary_10k_percent": 51.5,
        "gripper_x2_10k_percent": 48.0,
        "stage_balanced_10k_percent": overall,
        "change_vs_ordinary_points": overall - 51.5,
    }]),
], ignore_index=True)

comparison.to_csv(
    EVAL_ROOT / "comparison.csv", index=False
)

display(comparison.round(2))
print("Total attempts:", total)
print("Results, logs, and videos:", EVAL_ROOT)
```
